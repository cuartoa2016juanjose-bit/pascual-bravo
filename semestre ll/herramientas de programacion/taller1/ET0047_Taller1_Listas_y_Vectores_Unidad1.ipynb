## 1. Listas en Python: qué son y cómo se crean

Una **lista** es una colección **ordenada** y **mutable** de elementos.

- Ordenada: cada elemento tiene una posición (índice).
- Mutable: puedes modificar, agregar o eliminar elementos.

**Creación de listas:**
- Literales: `[]`
- A partir de otras estructuras: `list(...)`


```python
# Lista de números
edades = [18, 20, 19, 22, 21]
print(edades)

# Lista de textos
nombres = ["Ana", "Luis", "Camila"]
print(nombres)

# Lista mixta (se puede, pero en ejercicios numéricos preferimos listas homogéneas)
mixta = [1, "hola", True]
print(mixta)
```

### 1.1 Índices, acceso y modificación

- El primer elemento está en índice `0`.
- El último elemento se puede acceder con `-1`.


```python
edades = [18, 20, 19, 22, 21]

print("Primer elemento:", edades[0])
print("Último elemento:", edades[-1])

# Modificar un elemento
edades[1] = 25
print("Lista modificada:", edades)
```

### 1.2 Rebanado (slicing)

`lista[inicio:fin:paso]`

- `inicio` incluido
- `fin` excluido

## 2. Operaciones comunes con listas (herramientas integradas)

Estas herramientas ya vienen en Python y se pueden usar directamente (sin escribir `def`):

- `len(lista)` → cantidad de elementos
- `sum(lista)` → suma (si son números)
- `min(lista)`, `max(lista)` → menor / mayor
- `sorted(lista)` → devuelve una nueva lista ordenada
- `lista.count(x)` → cuántas veces aparece `x`
- `lista.index(x)` → posición de la primera aparición de `x`

Ejemplos:


```python
pesos = [72.5, 80.2, 65.0, 90.1, 77.7]

print("n =", len(pesos))
print("suma =", sum(pesos))
print("promedio =", mean(pesos))
print("mínimo =", min(pesos))
print("máximo =", max(pesos))

print("ordenada (copia) =", sorted(pesos))
print("original =", pesos)
```

## 3. Métodos de lista (lo más usado)

- `append(x)`: agrega al final
- `extend(iterable)`: agrega varios
- `insert(i, x)`: inserta en índice `i`
- `remove(x)`: elimina primera ocurrencia de `x`
- `pop(i)`: elimina y devuelve el elemento del índice `i` (por defecto el último)
- `count(x)`: cuántas veces aparece `x`
- `index(x)`: índice de la primera ocurrencia de `x`



<div style="text-align:center; margin:18px 0;">
  <img src="https://pbs.twimg.com/media/FSmDyTDX0BANJ5A.png" alt="meme gato" style="max-width:520px; width:90%; height:auto; border-radius:14px; box-shadow:0 6px 18px rgba(0,0,0,0.15);"/>
  <div style="margin-top:10px; font-size:14px; color:#444;"><em>El momento en que descubres `.append()` y `.pop()` y tu lista nunca vuelve a ser la misma 🧪</em></div>
</div>


```python
v = [3, 7, 6]

v.append(2)
print("append:", v)

v.extend([9, 8])
print("extend:", v)

v.insert(1, 100)
print("insert:", v)

print("count(7):", v.count(7))
print("index(6):", v.index(6))

ultimo = v.pop()
print("pop():", ultimo, "->", v)

v.remove(100)
print("remove(100):", v)
```

## 4. Recorrer listas con `for`

Patrones típicos:
- Recorrer elementos
- Recorrer índices con `range(len(lista))`
- Recorrer índice y elemento con `enumerate(lista)`



<div style="text-align:center; margin:18px 0;">
  <img src="https://preview.redd.it/meu-gato-%C3%A9-praticamente-o-gato-do-meme-v0-n9dleoj8dgfc1.jpg?width=640&crop=smart&auto=webp&s=52de2578fe227e639db822d551bba33c6d28a2f1" alt="meme gato" style="max-width:520px; width:90%; height:auto; border-radius:14px; box-shadow:0 6px 18px rgba(0,0,0,0.15);"/>
  <div style="margin-top:10px; font-size:14px; color:#444;"><em>Cuando el `for` recorre la lista y todo sale perfecto a la primera (casi nunca) 😺</em></div>
</div>


```python
v = [10, 20, 30]

print("Recorrer elementos:")
for x in v:
    print(x)

print("\nRecorrer índices:")
for i in range(len(v)):
    print(i, v[i])

print("\nEnumerate:")
for i, x in enumerate(v):
    print(i, x)
```

## 5. Transformaciones: comprensiones de lista (list comprehensions)

Forma compacta de construir listas:

`nueva = [expresión for elemento in lista if condición]`


```python
v = [1, 2, 3, 4, 5, 6]

cuadrados = [x**2 for x in v]
pares = [x for x in v if x % 2 == 0]
dobles_pares = [2*x for x in v if x % 2 == 0]

print("cuadrados:", cuadrados)
print("pares:", pares)
print("dobles_pares:", dobles_pares)
```

### 7.3 Aleatoriedad: barajar (idea del ejercicio 23)

`random.shuffle(lista)` baraja **en el lugar**.


```python
v = [3, 7, 6, 2, 9, 8]
w = v.copy()
random.shuffle(w)
print("v (original):", v)
print("w (barajada):", w)

coinciden = sum(1 for i in range(len(v)) if v[i] == w[i])
print("Coincidencias en la misma posición:", coinciden)
```
