## 1) ¿Qué es un string?

Un **string** es una secuencia de caracteres. En Python se escribe entre comillas:

- `'Hola'`
- `"Hola"`
- `'''texto\nmultilínea'''`

Los strings son **inmutables**: no puedes cambiar un carácter “en sitio” (en vez de eso, creas un string nuevo).


```python
# Ejecuta esta celda
s1 = 'Hola'
s2 = "Mundo"
s3 = '''Esto es
un texto
multilínea'''
print(s1, s2)
print(s3)
```

## 2) Longitud, índices y slicing

- `len(texto)` → número de caracteres.
- Índices: `texto[0]` es el primer carácter.
- Índices negativos: `texto[-1]` es el último carácter.
- Slicing: `texto[inicio:fin:paso]` (el `fin` no se incluye).


```python
texto = "Programación"
print("Texto:", texto)
print("Longitud:", len(texto))
print("Primer carácter:", texto[0])
print("Último carácter:", texto[-1])

print("Slice [0:4]:", texto[0:4])
print("Slice [5:]:", texto[5:])
print("Slice [::2] (saltando de 2):", texto[::2])
print("Reverso [::-1]:", texto[::-1])
```

## 3) Operaciones comunes con strings

- Concatenación: `a + b`
- Repetición: `a * n`
- Pertenencia: `'a' in texto`
- Comparación (alfabética): `"ana" < "beto"`


```python
a = "Hola"
b = "Python"

print(a + " " + b)
print("ha" * 4)

frase = "la vida es bella"
print("'vida' in frase:", "vida" in frase)
print("'V' in frase:", "V" in frase)
```

## 4) Métodos esenciales de strings

Piensa en los métodos como “herramientas” para transformar o analizar texto.

| Categoría | Métodos útiles |
|---|---|
| Normalizar | `lower()`, `upper()`, `title()`, `capitalize()` |
| Limpiar espacios | `strip()`, `lstrip()`, `rstrip()` |
| Reemplazar | `replace()` |
| Buscar | `find()`, `count()` |
| Separar / Unir | `split()`, `'sep'.join(lista)` |
| Validar | `isalpha()`, `isdigit()`, `isalnum()` |
| Prefijo/Sufijo | `startswith()`, `endswith()` |

> Nota: `find()` devuelve **-1** si no encuentra el texto.


```python
t = "   Hola, MUNDO. Bienvenidos a Python  "
print("Original:", repr(t))

t2 = t.strip()
print("strip():", repr(t2))

print("lower():", t2.lower())
print("upper():", t2.upper())

print("replace('Python','PY'):", t2.replace("Python", "PY"))

print("count('o'):", t2.count("o"))
print("find('MUNDO'):", t2.find("MUNDO"))
print("startswith('Hola'):", t2.startswith("Hola"))
print("endswith('  '):", t2.endswith("  "))

partes = t2.split()
print("split():", partes)
print("join():", "-".join(partes))
```

## 5) Formateo de texto (f-strings)

Una forma cómoda y clara de construir mensajes es con **f-strings**:


```python
producto = "Teclado"
precio = 129900
cantidad = 2

total = precio * cantidad
print(f"Producto: {producto}")
print(f"Cantidad: {cantidad}")
print(f"Total a pagar: ${total:,}")
```

## 6) Laboratorio guiado: normalizar y validar un texto

**Meta:** tomar un texto con espacios y mayúsculas, limpiarlo, y extraer datos.

> Ejecuta, entiende, luego modifica.


```python
entrada = "  CC:  1020451835   "
# Paso 1: limpiar espacios
limpio = entrada.strip()

# Paso 2: normalizar a mayúsculas (para comparar fácil)
limpio = limpio.upper()

# Paso 3: verificar si empieza con "CC:"
print("Texto limpio:", repr(limpio))
print("¿Empieza con CC?:", limpio.startswith("CC:"))

# Paso 4: extraer lo que está después de "CC:"
dato = limpio.replace("CC:", "").strip()

# Paso 5: validar si es numérico
print("Dato extraído:", dato)
print("¿Es número?:", dato.isdigit())
```
